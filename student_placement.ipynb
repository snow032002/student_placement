# Visualization

```python
plt.figure(figsize=(8,5))

plt.hist(df['salary_lpa'])

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Count")

plt.show()
```

```python
branch_salary = df.groupby('branch')['salary_lpa'].mean()

branch_salary.plot(kind='bar')

plt.title("Salary By Branch")
plt.show()
```

```python
company_salary = df.groupby('company_type')['salary_lpa'].mean()

company_salary.plot(kind='bar')

plt.title("Salary By Company Type")
plt.show()
```

```python
sns.scatterplot(x='internships_completed',y='salary_lpa',data=df)

plt.show()
```

```python
sns.boxplot(y=df['salary_lpa'])

plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap( df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

# Key Insights

- Placement rates vary significantly across academic branches and student profiles.

- Students with higher internship participation, stronger project portfolios, and better technical skills tend to receive more job offers.

- Certain company types offer higher salary packages and recruit students from specific branches more frequently.

- Communication skills, aptitude scores, and resume quality play an important role in placement success.

- Work mode preferences (Remote, Hybrid, Onsite) differ across company types and student groups.

- Skill gaps identified between placed and non-placed students highlight areas requiring additional training and development.

- Internship experience shows a positive relationship with employability and career opportunities.

# Conclusion

##### The Placement Intelligence Dashboard highlights the critical factors that drive student employability and placement success. The analysis demonstrates that a combination of academic performance, technical proficiency, practical experience, and communication skills significantly influences hiring outcomes. These insights can help recruiters identify high-potential candidates, assist institutions in improving training programs, and support students in developing the skills required for successful career opportunities.
